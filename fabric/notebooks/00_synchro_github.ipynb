# Synchronisation GitHub → Fabric

Met à jour les notebooks de l'espace de travail Fabric à partir des fichiers .ipynb
d'une branche du dépôt GitHub.

L'intégration Git native de Fabric avec GitHub n'est pas activée sur notre tenant :
ce notebook en tient lieu, dans un seul sens (le dépôt fait foi). Une modification
faite directement dans Fabric sur un notebook synchronisé est écrasée au prochain
lancement : modifier le code dans le dépôt, puis relancer la synchronisation.

Les notebooks absents du dossier du dépôt (ex. créés par un coéquipier dans Fabric)
ne sont jamais touchés. Le notebook qui lance la synchronisation doit figurer dans EXCLURE :
Fabric interdit à un notebook de réécrire sa propre définition.

Ce notebook exécute la dernière version de `fabric/synchro_github.py` sur la branche choisie. Il ne se met jamais à jour lui-même (Fabric l'interdit) : il est exclu de la synchronisation.

```python
DEPOT = "meardagoat/hackathon-mobility-citiz"
BRANCHE = "dev"  # branche à déployer dans Fabric (dev = intégration)
DOSSIER = "fabric/notebooks"
EXCLURE = "00_synchro_github"  # notebooks à ne pas écraser, séparés par des virgules
LAKEHOUSE_PAR_DEFAUT = "LakeHouseZ"  # si le notebook n'a pas de Lakehouse attaché
```

```python
import requests
exec(requests.get(f"https://raw.githubusercontent.com/{DEPOT}/{BRANCHE}/fabric/synchro_github.py", timeout=30).text)
```
